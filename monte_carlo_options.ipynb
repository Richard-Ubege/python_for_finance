```python
#Random walk

def random_walk(n, p):
    S_n = [0]
    for i in range(1, n):
        r = np.random.uniform()
        if r < p:
            S_n.append(S_n[i-1] + 1)
        else:
           S_n.append(S_n[i-1] - 1) 

    return S_n
```

```python

n = 100
x = np.linspace(0, n, n)
p = 0.4
S_n = random_walk(n, p)
plt.plot(x, S_n)
```
